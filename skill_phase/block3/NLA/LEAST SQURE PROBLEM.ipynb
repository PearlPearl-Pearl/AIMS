```julia
using Plots
using Statistics
using LinearAlgebra
```

```julia
# Step 1: Generate synthetic data
n = 20
x = 10 .* rand(n)
y = 3 .+ 2 .* x .+ randn(n)  # y = 3 + 2x + noise
```

```julia
# Step 2: Fit using least squares (linear regression)
X = [ones(n) x]             # Design matrix
β = X \ y                   # Least squares solution (intercept, slope)
```

```julia
# Predicted values
ŷ = X * β
```

```julia
# Step 3: Compute residuals
residuals = y - ŷ
rss = sum(residuals .^ 2)   # Residual sum of squares
```

```julia
# Step 4: Plot
scatter(x, y, label="Data points", legend=:topleft)
plot!(x, ŷ, label="Least squares fit", lw=2)
title!("Linear Regression Fit")
xlabel!("x")
ylabel!("y")
```

```julia
# Step 5: Print results
println("Fitted model: y = $(round(β[1], digits=2)) + $(round(β[2], digits=2)) * x")
println("Residuals: ", residuals)
println("Residual Sum of Squares (RSS): ", round(rss, digits=4))
```
