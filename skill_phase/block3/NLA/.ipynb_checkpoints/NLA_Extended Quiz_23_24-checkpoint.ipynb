**Exercise 1**

Cosider the system

\begin{equation}
    \begin{cases}
        2x+y=z+8\\
        -3x=y-2z-11\\ \label{eq1} \tag{1}
        -2x+y+2z=-3,
    \end{cases}
\end{equation}
write it in the form of $Ax=b$.

(a) Use **LU factorization** to solve the system \eqref{eq1} (*use your own forward and backward substituion function code, to be submitted*)

$\begin{bmatrix}2 & 1 & -1\\ 3 & 1 & -2\\ -2 & 1 & 2\end{bmatrix}\begin{bmatrix}x_1\\x_2\\x_3\end{bmatrix}=\begin{bmatrix}8\\11\\-3\end{bmatrix}$

```julia
using LinearAlgebra
```

```julia
function forward_sub(A,b)
    n = size(A)[1]
    x = zeros(n)
    
    for i in 1:n
        for j in 1:(i-1)
            b[i] = b[i]-(A[j,j]*x[j])        
        end
        x[i] = b[i]/A[i,i]
    end
    return x
end


function backward_sub(A,b)
    n = size(A)[1]
    x = zeros(n)
    
    for i in n:-1:1
        for j in (i+1):n
            b[i] = b[i]-(A[j,j]*x[j])        
        end
        x[i] = b[i]/A[i,i]
    end
    return x

end
```

```julia
A = [2 1 -1; 3 1 -2; -2 1 2]
b = [8; 11; -3]
L, U = LinearAlgebra.lu(A)
```

```julia
L
```

```julia
y = forward_sub(L, b)
```

```julia
x = backward_sub(U, y)
```

```julia
function my_qr(A, b)
    c = A[1, 1]/LinearAlgebra.norm(A[:,1], 2)
    s = A[2, 1]/LinearAlgebra.norm(A[:,1], 2)
    
    Q = [c -s; s c]
    Qprime = Q'
    
    R = Qprime*A
    
    x = R\(Qprime*b)
    return x
end
```

```julia
A = [1 2; 1 3]
b = [1; 2]
```

```julia
my_qr(A, b)
```

```julia
function householder_reflector(x)
    """This is a function that takes in an input x and zeros out all but the first entry"""
    x = reshape(x, length(x), 1) #First we make sure the input vector is a column vector
    n = length(x) #We generate the length of the column vector and call it n
    norm_x = LinearAlgebra.norm(x, 2) #We calculate the norm of the input vector
    y = zeros(n) #We initialize the y vector with zeros
    sign_x = sign(x[1]) #We get the sign of the first element of the column vector
    y[1] = sign_x*norm_x #We get the correct value of the first element of the y vector
    
    I = Matrix(1LinearAlgebra.I, n, n) #We initialize the identity matrix of order n
    
    
    u = x-y #We calculate the value of u
    γ = 2/((LinearAlgebra.norm(u,2))^2) #We generate the value of γ to use in the equation for the householder reflector 
    
    H = I - γ*(u*u') #We use the general formula for generating the householder reflector to get H
    return H #We return H
end
```

```julia
x = [1 2 3 4]
```

```julia
householder_reflector(x)*x'
```

```julia
householder_reflector(x)
```
