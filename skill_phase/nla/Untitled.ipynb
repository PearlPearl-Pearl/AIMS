```julia
using LinearAlgebra
```

```julia
function lsq_n(A, b)
    Aprime = A'
    B = Aprime*A
    bhat = Aprime*b
    R = cholesky(B).U
    y = R'\bhat
    x = R\y
    return x
end
```

```julia
A = [1 3 1; 1 3 7; 1 -1 -9; 1 -1 2]
b = [1; 1; 1; 1]

lsq_n(A, b)
```

```julia
function qr_decomposition(A, b)
    B  = LinearAlgebra.qr(A)
    R = collect(B.R)
    Q = collect(B.Q)
    zeros = [0 0 0]
    R = [R; zeros]
    
    
    y = Q\b
    x = R\y
    
    return x

end
```

```julia
A = [1 3 1; 1 3 7; 1 -1 -4; 1 -1 2]

qr_decomposition(A, b)
# LinearAlgebra.qr(A)
# F = LinearAlgebra.qr(A)
# collect(F.Q)
# collect(F.R)
```

```julia
function qr_factorization(A, b)
    B  = LinearAlgebra.qr(A)
    R = Matrix(B.R)
    Q = Matrix(B.Q)
    
    y = Q\b
    x = R\y
    return x
end
```

```julia
qr_factorization(A,b)
```

```julia
function svd_decomposition(A, b)
    a_svd = svd(A)
    
    U = a_svd.U
    Vt = a_svd.Vt
    S = diagm(a_svd.S)
    
    x = (U*S*Vt)\b
    return x

end
```

```julia
svd_decomposition(A,b)
```
