## Introduction

<span style='font-size:16px'>The advancements in technology in recent years have only increased the amount and complexity of data available for use. From satellite imagery to social media interactions, data has become the backbone of modern decision-making. As a result, the ability to manage, preprocess, and derive meaningful insights from data has become essential across both scientific research and industrial applications.
<br>
<br>
In particular, text data, which is often high-dimensional and unstructured, presents unique challenges. One such challenge being the conversion of the text data into a numeric representation since computers can only store data, at a low level, in bits. For example, while a human reader can easily understand that the words <i>“car”</i> and <i>“automobile”</i> are semantically similar, a computer initially sees them as completely different sequences of characters. Without some additional preprocessing to represent the text in a form that computers can interpret, it becomes impossible for algorithms to recognize or quantify such relationships.
<br>
<br>
This is where <strong>linear algebra</strong> comes in. Linear algebra allows us to study the structure of different spaces, and how the elements of those spaces can be transformed within the same space or into another space through <em>linear transformations</em>. Put simply, linear transformations allow us to convert the unstructured space of text into structured numerical representations that can be processed and analyzed.
<br>
<br>
Once text data has been converted into a numeric form through linear transformations, we can apply the <em>Singular Value Decomposition (SVD)</em>. SVD is a technique that allows us to simplify complex transformations by identifying and retaining only the most important patterns in the data, while filtering out noise or less relevant information. In this way, SVD helps us focus on the key structure within the data, making analysis more efficient and meaningful.
In essence, SVD provides a practical bridge from the unstructured text data to the structured numerical data on which analysis can be done, motivating its central role in this work.
<br>
<br>
In this project, we focus on using SVD within the framework of Latent Semantic Analysis (LSA) to analyze text data. By constructing a weighted term-document matrix and decomposing it with SVD, we can identify latent relationships between words and documents, capturing semantic structure beyond what is visible on the surface. This approach enables tasks such as document clustering, information retrieval, and semantic similarity measurement, demonstrating the power of linear algebraic techniques in extracting meaningful patterns from high-dimensional text data.
</span>
