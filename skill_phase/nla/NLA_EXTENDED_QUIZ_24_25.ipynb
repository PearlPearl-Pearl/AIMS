**Exercise 1**

Let $$A = 
\begin{pmatrix}
2 & 16 & 17 \\
1 & 2 & 3 \\
2 & 8 & 11 
\end{pmatrix}
$$

1\. Find a decomposition of the form $PA = LU$ for the matrix $A$.

2\. Solve the system $Ax = [51,\; 6, \; 19]^{T}$ using part 1.

```julia
using LinearAlgebra
```

```julia
A = [2 16 17; 1 2 3; 2 8 11]
```

```julia
L, U = LinearAlgebra.lu(A)
```

```julia
P = [1 0 0; 0 0 1; 0 1 0]
```

```julia
L*U == P*A
```

```julia
b = [51; 6; 19]
```

```julia
x = (P'*L*U)\b
```
